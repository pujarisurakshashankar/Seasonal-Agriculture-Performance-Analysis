#Separate Numerical and Categorical Columns

This separates the dataset columns into two groups. Numerical columns contain numbers, while categorical columns contain text or categories. This helps us choose the correct analysis and charts.

```python
numerical_columns = df.select_dtypes(include=np.number).columns
categorical_columns = df.select_dtypes(include='object').columns

print("Numerical columns:")
print(numerical_columns.tolist())

print("\nCategorical columns:")
print(categorical_columns.tolist())
```

#Handle Missing Numerical Values

This step fills missing values in numerical columns with the median value of each column. The median is used because it is less affected by very high or very low values.

```python
for column in numerical_columns:
    df[column] = df[column].fillna(df[column].median())

print("Missing numerical values handled successfully!")
```

# Handle Missing Categorical Values

This fills missing values in categorical columns with the most frequently occurring value in each column. The most frequently occurring value is called the mode.

```python
for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])

print("Missing categorical values handled successfully!")
```

#Remove Duplicate Records

Duplicate records are repeated rows in the dataset. This step removes duplicate rows so that the same record is not counted more than once during analysis.

```python
df = df.drop_duplicates()

print("Duplicate records removed successfully!")
```

# Verify the Final Dataset

This checks the final number of rows and columns, the total missing values, and the total duplicate records after data cleaning.

```python
print("Final dataset shape:", df.shape)

print("\nTotal missing values after cleaning:")
print(df.isnull().sum().sum())

print("\nTotal duplicate records after cleaning:")
print(df.duplicated().sum())
```

#Select Columns for Correlation Analysis

This selects the important numerical columns that will be used for correlation analysis. Correlation helps us understand the relationship between rainfall, temperature, humidity, soil moisture, water usage, crop yield, and profit.

The selected columns will be used to create a correlation matrix and heatmap.

```python
correlation_columns = [
    'Rainfall_mm',
    'Avg_Temperature_C',
    'Humidity_pct',
    'Soil_Moisture_pct',
    'Water_Used_m3',
    'Yield_Tonnes_Ha',
    'Profit_INR'
]

correlation_matrix = df[correlation_columns].corr()

print("Correlation matrix:")
display(correlation_matrix)
```

#Create a Correlation Heatmap

This  creates a heatmap to show the correlation between environmental conditions and agricultural performance.

Correlation values range from -1 to +1:

- A value close to +1 shows a strong positive relationship.
- A value close to -1 shows a strong negative relationship.
- A value close to 0 shows little or no relationship.

The heatmap uses different colors and displays the correlation values inside each cell.

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Between Environmental Conditions and Performance')

plt.show()
```

# Analyze Yield by State and Season

This  calculates the average crop yield for each combination of state and agricultural season.

It helps us compare the performance of different states across different seasons and identify which state-season combination produces the highest average yield.

```python
state_season_yield = df.groupby(['State', 'Season'])[
    'Yield_Tonnes_Ha'
].mean().reset_index()

print("Average yield by state and season:")
display(state_season_yield)
```

#Create a State and Season Yield Heatmap

This  creates a pivot table and heatmap to compare the average crop yield of different states across different agricultural seasons.

The heatmap makes it easier to identify:

- States with higher average yield.
- Seasons in which yield is higher or lower.
- State-season combinations with the best performance.

The values are displayed up to two decimal places.

```python
state_season_pivot = df.pivot_table(
    values='Yield_Tonnes_Ha',
    index='State',
    columns='Season',
    aggfunc='mean'
)

plt.figure(figsize=(10, 6))

sns.heatmap(
    state_season_pivot,
    annot=True,
    fmt='.2f',
    cmap='YlGnBu'
)

plt.title('Average Yield by State and Season')
plt.xlabel('Season')
plt.ylabel('State')

plt.show()
```

#Create a Seasonal Performance Summary

This calculates the average yield, revenue, profit, and water usage for each agricultural season.

It helps us compare the overall performance of different seasons and understand how much water is used to achieve the agricultural results.

```python
season_summary = df.groupby('Season')[
    [
        'Yield_Tonnes_Ha',
        'Revenue_INR',
        'Profit_INR',
        'Water_Used_m3'
    ]
].mean()

print("Seasonal performance summary:")
display(season_summary)
```

# Identify the Best Performing Season

This identifies the season with the highest average crop yield and the highest average profit.

It helps us understand which season performs best in terms of agricultural productivity and financial returns.

```python
best_yield_season = season_summary['Yield_Tonnes_Ha'].idxmax()
best_profit_season = season_summary['Profit_INR'].idxmax()

print("Season with the highest average yield:", best_yield_season)
print("Season with the highest average profit:", best_profit_season)
```

#Create the Final Seasonal Comparison

This  creates the final comparison of agricultural seasons using average crop yield, revenue, and profit.

The values are rounded to two decimal places so that the results are easier to read and compare.

This table provides a clear summary of the overall seasonal performance.

```python
final_comparison = df.groupby('Season')[
    [
        'Yield_Tonnes_Ha',
        'Revenue_INR',
        'Profit_INR'
    ]
].mean().round(2)

print("Final seasonal comparison:")
display(final_comparison)
```

# Identify the Best Seasonal Performance

This identifies the season with the highest average yield, revenue, and profit from the final seasonal comparison.

The results will help us prepare the conclusion of the agriculture performance analysis project.

```python
best_yield = final_comparison['Yield_Tonnes_Ha'].idxmax()
best_revenue = final_comparison['Revenue_INR'].idxmax()
best_profit = final_comparison['Profit_INR'].idxmax()

print("Conclusion:")
print("The season with the highest average yield is", best_yield)
print("The season with the highest average revenue is", best_revenue)
print("The season with the highest average profit is", best_profit)
```
